1. Create two variables a and b with values 10 and 5. Print their sum, difference, product, and division.

```python
a = 10
b = 5

total = a + b
diff = a - b
prod = a * b
div = a / b
print("total is: ", total, "\n",
"difference is: ", diff, "\n",
"The product is: ", prod, "\n",
"The division is: ", div, "\n",
"========= Done ! =============")
```

2. Store your age in a variable and print: “I am X years old”.

```python
my_age = 37
# Simplest way
print("I am", my_age, "years old")

# Using string formatting and placeholders
print("I am {} years old".format(my_age))

# Preffered way using f-strings. Note: f-strings work differently depending on the - 
# types of quotes used e.g., "" and """ """
print(f"I am {my_age} years old")
```

3. Swap two variables without using a third variable.

```python
a = 22
b = 34

# test
print(f"a is {a} and b is {b}")
print("======= Swapping Variables =========")

a,b = b, a

# test
print(f"a is now {a} and b is {b}")
```

4. Convert temperature from Celsius to Fahrenheit.

    `` Fahrenheit =(Celsius * 1.8) + 32 ``

```python
celsius_temp = int(input("Enter the Celsius temperature: "))
fahren_temp = (celsius_temp * 1.8) + 32

print(f"Conversion finished! {celsius_temp} Degree Celsius is {fahren_temp} Fahrenheit")
```

6. Take a number and check if it is even or odd.

```python
my_num1 = int(input("Enter a number: "))

# Check!
print(f"You have entered the number {my_num1}")

if my_num1 % 2 == 0 and my_num1 != 0:
    print("This number is Even")
else:
    print("This number is Odd!")
```

7. Given three numbers, print the largest one.

```python
num_a = int(input("Enter the first number: "))
num_b = int(input("Enter the second number: "))
num_c = int(input("Enter the third number: "))

if num_a >= num_b and num_a >= num_c:
    print(f"The largest number is {num_a}")

elif num_b >= num_a and num_b >= num_c:
    print(f"The largest number is {num_b}")

elif num_c >= num_a and num_c >= num_b:
    print(f"The largest number is {num_c}")
```

10. Given a number, print all numbers from 1 up to that number.

```python
given_num = int(input("Enter any Number: "))

if given_num <= 1:
    print("That number is too small, try numbers bigger than 2")

elif given_num >= 2:
    for nums in range(1, given_num + 1):
        print(nums)
```

11. Print the sum of numbers from 1 to N.

```python
num_n = int(input("Enter the Nth value: "))
total = 0
print(f"The nth number provided is {num_n}")

for nums in range(1, num_n + 1):
        total = total + nums

print(f"The total of nums from 1 to {num_n} is {total}")

```
