A configuration reader needs to skip blank lines before reading the first useful setting.

Write `first_nonempty_line(text: str) -> str | None` that returns the first line containing non-whitespace characters, stripped of surrounding whitespace, or None if no such line exists.

```python
file_path = "./read_file.txt"

def first_nonempty_line(filepath: str) -> str | None:
# use the 'with' statement to read and close the file automatically
    with open(file_path) as f:
        # loop through the lines of the file
        for line in f:
            # strip empty lines at the beginning and end of a line
            line = line.strip()

            # if a line is encountered
            if line:
                return line

    return None

first_nonempty_line(file_path)
```

A reading tool wants to show the length of each word in a sentence for a beginner vocabulary exercise.

Write a function `word_lengths(sentence: str) -> dict[str, int]` that returns a dictionary mapping each whitespace-separated word to its length. Ignore empty input and repeated words should use the same key.

```python
def word_lengths(sentence: str) -> dict[str, int]:
    if len(sentence) != 0:
        # create a list of the parts of the sentence
        elements = sentence.split()

        # create an empty dict
        wordcount_store = {}

        # loop through the elements of the sentence
        for element in elements:
            value = len(element)
            key = element.lower()
            # update the empty dict
            wordcount_store.update({key : value})
        return wordcount_store

    else:
        print("Empty string provided!")
    return {}
    

my_str = "And this distinction becomes particularly useful in programming and NLP, because you'll encounter terms like character And SAS which becomes"
my_str2 = ""
print(word_lengths(my_str))
print(word_lengths(my_str2))
```

A classroom game needs to convert each turn number into its FizzBuzz label according to its divisibility rules.

Write a function `fizz_buzz_value(number: int) -> str` that returns `FizzBuzz` when number is divisible by 3 and 5, `Fizz` when divisible only by 3, `Buzz` when divisible only by 5, and the number as a string otherwise.

```python
class FizzBuzz:

    def __init__(self, num):
        self.num = num

    def fizz_buzz_value(self) -> str:
        """
        A method that operates onthe given number.
        """
        # check if num is divisible by 3, 3 and 5, or 5
        if (self.num % 3 == 0 and self.num % 5 == 0):
            return "FizzBuzz"

        elif (self.num % 3 == 0 and self.num % 5 !=0):
            return "Fizz"
        
        elif (self.num % 5 == 0 and self.num % 3 != 0):
            return "Buzz"

        return str(self.num)

fizzbee = FizzBuzz(33)
print(fizzbee.fizz_buzz_value())
```
