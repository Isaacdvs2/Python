1. Simple Greeting App
* Procedural Approach: Store name in variable, print greeting using functions
* OOP ApproacH: Create Person class with greet() method

```python
# 1. Procedural multiple approaches
name = "Isaac"

def greet():
    print(f"Hello {name}, from main greet function")

print(greet())

def greet_v1(name_1):
    print(f"Hello, {name_1}, from greet_v1 function")

greet_v1(name)
name2 = "John"
greet_v1(name2)

def greet_v2():
    name_v2 = input("Enter Your Name: ")
    print(f"Hi {name_v2}, from greet_v2 function")

greet_v2()

# 2. OOp approach
class person():
    #constructor
    def __init__(self):
        self.__name = name

    # The greeting method
    def greetings2You(self):
        self.name = input("Enter Your Name: ")
        print(f"Dumela / Hi,  {self.name}")

new_greetings = person()
new_greetings.greetings2You()


```

2. Student Marks Tracker
* Procedural Approach: Use lists/dicts to store names + marks
* OOP: Student class with attributes + display_result()

```python
# 1. Procedural approach

# Create a function that accepts the student name and marks and store them as a dictionary.
# Simple
student_marks = {}

def store_class_marks(name, marks):
    student_marks.update({name: marks})

name = "Isaac"
marks = 40
name_1 = "Brady"
marks_1 = 300
store_class_marks(name, marks)
store_class_marks(name_1, marks_1)
# #verify
print(student_marks)
print("============ Simple Fn Done ! ================", "\n")

# # V1
student_marks_v1 = {}
def store_class_marks_v1():
    name_v1 = input("Enter Your Student Name: ")
    marks_v1 = int(input("Enter the student's Marks: "))
    student_marks_v1.update({name_v1 : marks_v1})

# # manually
store_class_marks_v1()
store_class_marks_v1()
store_class_marks_v1()

# # or just Loop
for i in range(5):
    store_class_marks_v1()

# verify
print(student_marks_v1)
print("============== V1 Fn Done! ================", "\n")

print(f"""
      ========= Starting the OOP Way! =================""")
# OOP Approach

class student_names_marks():
    # inititate the constructor
    def __init__(self):
        self.__studentMarks = {}

    def store_studentMarks(self):
        name_2 = input("Enter the student's name: ")
        marks_2 = int(input("Enter the student's marks: "))
        self.__studentMarks.update({name_2 : marks_2})

    def display_students(self):    
        return self.__studentMarks

# Manually !!!
semester = student_names_marks()
semester_1 = semester.store_studentMarks()
semester_2 = semester.store_studentMarks()
semester_3 = semester.store_studentMarks()
# verify
print(f"The results for the first 3 semesters are as follows \n {semester.display_students()}")

# Just loop
for i in range(3):
    semester.store_studentMarks()

print(f"The results for the Year are as follows \n {semester.display_students()}")

# Even better -> Export the results to a csv file (write to the master csv file)
# Even Better -> allow the user to upload their csv, process it and append to the master -
# database (Let the user decide whether to input or to upload)


```

3. Grocery List Manager
* Procedural Approach: List of items + functions add/remove
* OOP: GroceryList class managing items internally

```python
# 1. Procedural

grocer_list = []

def add_stuff():
    add = input("Type the item to add to the list: ")
    # check if the items is not already added
    if add in grocer_list:
        print(f"That item has already been added \n")
    else:
        grocer_list.append(add)
        
    return grocer_list

def remove_stuff():
    # removes the last item from the list
    grocer_list.pop()
    return grocer_list


# Ideas
# -> Order items.
# -> Let the grocery list be a dict with the number of items to buy per item
# -> display the list and the user must decide on the items to remove.
# -> If the user adds an existing item, increase its number only in the dictionary
# -> Handle errors

grocer_list_v1 = {} # A dict also automatically orders items

def review_list():
    # Display the list to the user, check if list has items first
    if len(grocer_list_v1) >= 1:
        print("Your Groceries List is as Follows \n")
        print(f"""Item        |       Count 
              --------------------------------""")
        for keys, values in grocer_list_v1.items():
            print(f"{values}    |       {keys}")
    else:
        print(" Your list is currently empty!")
    return

def add_stuff_v1(num, item):
    # review the list before adding anything
    review_list()
    # convert the item names to lower case and remove trailing spaces before adding them
    item = item.lower().strip()
    # check if items exists in the dict. If they do, ask the user how
    # many more to add.
    if item in grocer_list_v1:
        print(f"""Item already on the list, as follows
        {grocer_list_v1.get(item)} X {grocer_list_v1[item]}
              """)
        add_more = int(input("How many more to add/remove: "))
        # Check if the user provided input
        if (add_more == ""):
            # if the user does not provide input, add zero to items total
            grocer_list_v1[item] = num + 0

        else:
            grocer_list_v1[item] = num + add_more
    # if its a new item simply add it
    else:
        grocer_list_v1[item] = num
    return 

def remove_stuff_v1(item):
    # review the list first
    review_list()

    # Ask to remove item completely
    to_beRemoved = input("Type the item to remove: ")
    # Convert the item typed to lower case
    to_beRemoved.lower().strip()
    # Check if the item is in the existing list
    if to_beRemoved in grocer_list_v1:
        grocer_list_v1.pop(to_beRemoved)
        
    else:
        print("Item not Found! Add it first")
    return

# 2. OOP Way


    
    


```
